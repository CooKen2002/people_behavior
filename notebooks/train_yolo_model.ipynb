```python
import yaml

yaml_path = "/content/dataset/data.yaml"

# Load current YAML configuration
with open(yaml_path, "r") as f:
    data = yaml.safe_load(f)

# Set the root path to Colab's dataset directory
data["path"] = "/content/dataset"

# Save updated config
with open(yaml_path, "w") as f:
    yaml.dump(data, f)

print("Updated data.yaml successfully!")
```

```python
# 1. Install Ultralytics
!pip install ultralytics -q

from ultralytics import YOLO

# 2. Load a base pre-trained model (YOLOv8 or YOLO11)
model = YOLO("yolo11n.pt")  # or "yolov8n.pt"

# 3. Start custom training
results = model.train(
    data="/content/dataset/data.yaml",
    epochs=50,  # Increase to 80-100 for better accuracy if needed
    imgsz=640,
    batch=16,
    name="classroom_detector",
)
```
